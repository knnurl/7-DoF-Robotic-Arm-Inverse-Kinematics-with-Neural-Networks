```python
# Functions to extract useful position data from the messy string above.

# Python3 code to demonstrate working of
# Extract String after Nth occurrence of K character
# Using list comprehension and enumerate()
def get_pose_ori(poses,i):
    # initializing string
    test_x = poses.iloc[i,1]

    # printing original string
    #print("The original string is : " + str(test_str))

    # initializing K
    K = 'x'

    # initializing N
    N = 17

    # list comprehension to get the indices of K character
    indices = [i for i, x in enumerate(test_x) if x == K]

    # extracting string after Nth occurrence of K character
    if len(indices) >= N:
        result_pose = test_x[indices[N-1]+1:]
        result_pose = result_pose[2:]
    #    print("The pose_ori :\n " + str(respose))
    #else:
    #    print("Error: K character not found N times in the string.")

    return result_pose

# Leave out the orientations as the are not going to be utilized
def get_poses(poses_ori):
    # Python3 code to demonstrate working of
    # Prefix extraction before specific character
    # Using rsplit()
    # initializing split character
    spl_char = '\nori'
    # printing original string
    #print("The original string is :"+ str(test_y))
    # Using rsplit()
    # Prefix extraction before specific character
    poses_xyz = poses_ori.rsplit(spl_char, 2)[0]
    return poses_xyz

# Split x from the rest and save it to a variable
def get_x_yz(poses_xyz):
    x = poses_xyz.split('\n  y: ')[0]
    yz = poses_xyz.split('\n  y: ')[1]
    return x, yz

# Split y and z from each other
def get_y_z(yz):
    y = yz.split('\n  z: ')[0]
    z = yz.split('\n  z: ')[1]
    return y, z

# Combine all functions in one function
def parse_xyz(poses,i):
    posori = get_pose_ori(poses,i)
    poses_xyz = get_poses(posori)
    x , yz = get_x_yz(poses_xyz)
    y, z = get_y_z(yz)
    return x, y, z
```

```python
df_clean_xyz = [parse_xyz(poses,i) for i in range(len(poses))]
```

```python
# Create a new dataframe only with Time column of angles dataset
df_angles_time = df_angles.iloc[:,0:1]

# Leave out columns of df_angle other than angles
df_angles = df_angles.iloc[:,6:13]

print(df_angles.columns)

# Rename columns
df_angles.rename(columns={'position_0': 'theta0',
                          'position_1': 'theta1',
                          'position_2': 'theta2',
                          'position_3': 'theta3',
                          'position_4': 'theta4',
                          'position_5': 'theta5',
                          'position_6': 'theta6'},
                          inplace=True, errors='raise')
df_angles
```

```python
# Concatenate the times and angles
df_angles_timed = pd.concat([df_angles_time,df_angles], axis=1)
df_angles_timed
```

```python
# Concatanate the 3D positions and joint angles
# only merge the samples whose values in Time columns are matching
# This allow the proper syncronisation of xyz and angle values
result_df = pd.merge(df_xyz_timed, df_angles_timed, on='Time', how='inner')
```

```python
final_dataset = result_df.rename(columns={0: 'x',
                                          1: 'y',
                                          2: 'z'}, errors='raise')
```
